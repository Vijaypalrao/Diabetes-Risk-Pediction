```python
# ============================================================
# CRAMER'S V FUNCTION
# ============================================================

from scipy.stats import chi2_contingency
import numpy as np

def cramers_v(x, y):

    contingency = pd.crosstab(x, y)

    chi2 = chi2_contingency(contingency)[0]

    n = contingency.sum().sum()

    r, c = contingency.shape

    return np.sqrt(chi2 / (n * (min(r-1, c-1))))
```

```python
# ============================================================
# ASSOCIATION OF FEATURES WITH TARGET
# ============================================================

association_results = []

for feature in df.columns:

    if feature in ["Age", "class"]:

        continue

    value = cramers_v(

        df[feature],

        df["class"]

    )

    association_results.append(

        [feature, value]

    )

association_df = pd.DataFrame(

    association_results,

    columns=[

        "Feature",

        "Cramer's V"

    ]

)

association_df = association_df.sort_values(

    by="Cramer's V",

    ascending=False

)

display(association_df)
```

```python
# ============================================================
# 6.1 SHAPIRO-WILK NORMALITY TEST
# ============================================================

print("=" * 80)
print("SHAPIRO-WILK NORMALITY TEST")
print("=" * 80)

statistic, p_value = shapiro(df["Age"])

print(f"Test Statistic : {statistic:.5f}")
print(f"P-value        : {p_value:.5f}")
```

```python
# ============================================================
# DECISION
# ============================================================

alpha = 0.05

print("=" * 80)
print("DECISION")
print("=" * 80)

if p_value > alpha:

    print("Result : Age follows a Normal Distribution.")

    print("Decision : Fail to Reject the Null Hypothesis.")

    print("Recommended Test : Independent t-test")

else:

    print("Result : Age does NOT follow a Normal Distribution.")

    print("Decision : Reject the Null Hypothesis.")

    print("Recommended Test : Mann-Whitney U Test")
```

```python
# ============================================================
# PHASE 6.1 SUMMARY
# ============================================================

print("=" * 80)
print("PHASE 6.1 SUMMARY")
print("=" * 80)

print(f"Shapiro-Wilk Statistic : {statistic:.5f}")

print(f"P-value                : {p_value:.5f}")

if p_value > alpha:

    print("\nConclusion : Age is Normally Distributed.")

    print("Next Step : Independent t-test")

else:

    print("\nConclusion : Age is NOT Normally Distributed.")

    print("Next Step : Mann-Whitney U Test")

print("\nSTATUS : PHASE 6.1 COMPLETED")
```

```python
# ============================================================
# IMPORT LIBRARY
# ============================================================

from scipy.stats import mannwhitneyu
```

```python
# ============================================================
# SEPARATE AGE BY DIABETES STATUS
# ============================================================

age_positive = df[df["class"] == "Positive"]["Age"]

age_negative = df[df["class"] == "Negative"]["Age"]

print("=" * 80)
print("GROUP INFORMATION")
print("=" * 80)

print(f"Positive Patients : {len(age_positive)}")

print(f"Negative Patients : {len(age_negative)}")
```

```python
# ============================================================
# MANN-WHITNEY U TEST
# ============================================================

u_statistic, p_value = mannwhitneyu(

    age_positive,

    age_negative,

    alternative="two-sided"

)

print("=" * 80)
print("MANN-WHITNEY U TEST")
print("=" * 80)

print(f"U Statistic : {u_statistic:.4f}")

print(f"P-value     : {p_value:.6f}")
```

```python
# ============================================================
# DECISION
# ============================================================

alpha = 0.05

print("=" * 80)
print("DECISION")
print("=" * 80)

if p_value < alpha:

    print("Reject the Null Hypothesis (H₀).")

    print("Conclusion : There is a statistically significant")

    print("difference in Age between Positive and Negative patients.")

else:

    print("Fail to Reject the Null Hypothesis (H₀).")

    print("Conclusion : No statistically significant")

    print("difference in Age between Positive and Negative patients.")
```

```python
# ============================================================
# EFFECT SIZE (RANK-BISERIAL CORRELATION)
# ============================================================

n1 = len(age_positive)

n2 = len(age_negative)

rank_biserial = 1 - (2 * u_statistic) / (n1 * n2)

print("=" * 80)
print("EFFECT SIZE")
print("=" * 80)

print(f"Rank-Biserial Correlation : {rank_biserial:.4f}")
```

```python
# ============================================================
# EFFECT SIZE INTERPRETATION
# ============================================================

abs_r = abs(rank_biserial)

print("=" * 80)
print("EFFECT SIZE INTERPRETATION")
print("=" * 80)

if abs_r < 0.10:

    print("Negligible Effect")

elif abs_r < 0.30:

    print("Small Effect")

elif abs_r < 0.50:

    print("Medium Effect")

else:

    print("Large Effect")
```

```python
# ============================================================
# CHI-SQUARE TEST
# ============================================================

results = []

for feature in categorical_features:

    contingency_table = pd.crosstab(

        df[feature],

        df["class"]

    )

    chi2, p_value, dof, expected = chi2_contingency(

        contingency_table

    )

    results.append({

        "Feature": feature,

        "Chi-Square": round(chi2, 4),

        "Degrees of Freedom": dof,

        "P-value": round(p_value, 6),

        "Significant": "Yes" if p_value < 0.05 else "No"

    })

chi_square_results = pd.DataFrame(results)

display(chi_square_results)
```

```python
# ============================================================
# SORT RESULTS
# ============================================================

chi_square_results = chi_square_results.sort_values(

    by="P-value"

)

display(chi_square_results)
```

```python
# ============================================================
# SIGNIFICANT FEATURES
# ============================================================

significant_features = chi_square_results[

    chi_square_results["Significant"] == "Yes"

]

print("=" * 80)
print("STATISTICALLY SIGNIFICANT FEATURES")
print("=" * 80)

display(significant_features)
```

```python
# ============================================================
# SUMMARY
# ============================================================

print("=" * 80)
print("CHI-SQUARE SUMMARY")
print("=" * 80)

print(f"Total Features Tested      : {len(categorical_features)}")

print(f"Significant Features       : {len(significant_features)}")

print(f"Non-significant Features   : {len(categorical_features)-len(significant_features)}")
```

```python
# ============================================================
# VISUALIZATION
# ============================================================

plt.figure(figsize=(10,7))

sns.barplot(

    data=chi_square_results,

    y="Feature",

    x="Chi-Square",

    palette="viridis"

)

plt.title("Chi-Square Statistic for Each Feature")

plt.xlabel("Chi-Square Value")

plt.ylabel("Feature")

plt.tight_layout()

plt.show()
```

```python
# ============================================================
# PHASE 24
# DIABETES PREDICTION FUNCTION
# ============================================================

def predict_diabetes(
    age,
    gender,
    polyuria,
    polydipsia,
    sudden_weight_loss,
    partial_paresis,
    itching,
    irritability,
    polyphagia,
    alopecia
):

    patient = pd.DataFrame({

        "Polyuria":[polyuria],

        "Polydipsia":[polydipsia],

        "Gender":[gender],

        "sudden weight loss":[sudden_weight_loss],

        "partial paresis":[partial_paresis],

        "Itching":[itching],

        "Irritability":[irritability],

        "Polyphagia":[polyphagia],

        "Alopecia":[alopecia],

        "Age":[age]

    })

    patient = patient[selected_features]

    prediction = model.predict(patient)[0]

    probability = model.predict_proba(patient)[0][1]

    probability_percent = probability * 100

    if prediction == 1:

        status = "Positive"

    else:

        status = "Negative"

    if probability_percent < 20:

        risk = "Very Low Risk"

    elif probability_percent < 40:

        risk = "Low Risk"

    elif probability_percent < 60:

        risk = "Moderate Risk"

    elif probability_percent < 80:

        risk = "High Risk"

    else:

        risk = "Very High Risk"

    print("="*80)
    print("DIABETES RISK PREDICTION")
    print("="*80)

    print(f"Prediction               : {status}")

    print(f"Probability of Diabetes  : {probability_percent:.2f}%")

    print(f"Risk Category            : {risk}")

    print("="*80)

    return {

        "Prediction": status,

        "Probability": probability_percent,

        "Risk Category": risk

    }
```
